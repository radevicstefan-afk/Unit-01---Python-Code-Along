### Slicing lists

A slice uses `[start:stop]`: it includes `start` and excludes `stop`.

- `values[1:4]` selects positions 1, 2, and 3.
- `values[:3]` selects the first three values.
- `values[2:]` selects from position 2 through the end.
- `values[::2]` selects every second value.

````python
daily_sales = [125.50, 98.00, 156.75, 130.25, 201.00]

print('First three days:', daily_sales[:3])
print('Middle values:', daily_sales[1:4])
print('From the third day onward:', daily_sales[2:])
print('Every second value:', daily_sales[::2])
````

### Updating lists

### Mutability and copying

````python
original_scores = [72, 85, 91]
same_scores = original_scores
copied_scores = original_scores.copy()

same_scores.append(100)
copied_scores.append(60)

print('Original list:', original_scores)
print('Same reference:', same_scores)
print('Independent copy:', copied_scores)
````

## 3. Iterating through lists

A `for` loop visits every item in a list. Choose a meaningful singular name for the current item. Use `enumerate()` when you need both the item and its position.

````python
daily_sales = [125.50, 98.00, 156.75, 130.25, 201.00]

for sale in daily_sales:
    if sale >= 150:
        print(f'{sale:.2f} is a high-sales day.')
    else:
        print(f'{sale:.2f} is below the high-sales threshold.')

print()
for day_number, sale in enumerate(daily_sales, start=1):
    print(f'Day {day_number}: {sale:.2f}')
````

## 4. Dictionaries: records with named fields

A dictionary maps a **key** to a **value**. Dictionaries are useful for individual records, such as a customer, product, learner, or API response.

```python
learner = {
    'name': 'Amina',
    'city': 'Herceg Novi',
    'completed_modules': 2
}
```

````python
learner = {
    'name': 'Amina',
    'city': 'Herceg Novi',
    'completed_modules': 2,
    'active': True
}

print(learner)
print('Name:', learner['name'])
print('Modules completed:', learner['completed_modules'])
print('Preferred language:', learner.get('preferred_language', 'Not recorded'))
````

### Accessing dictionary values safely

`record['key']` is direct access. It raises a `KeyError` when the key does not exist.

`record.get('key')` returns `None` when the key does not exist. `record.get('key', default_value)` returns your supplied default instead. Use `.get()` for optional fields in external or incomplete data.

````python
product = {
    'product_id': 101,
    'name': 'Wireless Keyboard',
    'price': 49.99
}

product['stock'] = 35
product['price'] = 44.99

print(product)
print('Category:', product.get('category', 'Uncategorised'))
````

### Iterating through dictionaries

- `dictionary.keys()` returns the keys.
- `dictionary.values()` returns the values.
- `dictionary.items()` returns key/value pairs and is the common choice when you need both.

````python
product = {
    'product_id': 101,
    'name': 'Wireless Keyboard',
    'price': 44.99,
    'stock': 35
}

for key, value in product.items():
    print(f'{key}: {value}')
````

## 5. Errors are useful evidence

An exception is Python’s way of reporting that it cannot continue with the current instruction. Do not hide errors blindly; read them and handle only problems you expect can happen.

| Error | Typical cause | Example |
|---|---|---|
| `ValueError` | Correct type, invalid value | `int('five')` |
| `KeyError` | Dictionary key does not exist | `record['phone']` |
| `IndexError` | List position does not exist | `values[10]` |
| `TypeError` | Incompatible operation or argument types | `'5' + 2` |
| `NameError` | A variable name is not defined | `print(total_sales)` before assignment |

A traceback shows where the error occurred. Read the final line first: it names the exception and provides a concise description.

## 6. Defensive exception handling

Use a `try` block around a small operation that may fail. Use specific `except` blocks to respond to expected error types.

```python
try:
    value = int(text_value)
except ValueError:
    print('Please enter a whole number.')
else:
    print('Conversion succeeded:', value)
finally:
    print('This runs whether or not an error occurred.')
```

### Important rule
Avoid a bare `except:` in beginner code. It can hide unrelated bugs, including programming mistakes. Catch the specific exception you expect.

````python
raw_quantity = '12'

try:
    quantity = int(raw_quantity)
except ValueError:
    print('Invalid quantity. Enter a whole number.')
else:
    print(f'Quantity accepted: {quantity}')
finally:
    print('Validation attempt complete.')
````

````python
raw_quantity = 'twelve'

try:
    quantity = int(raw_quantity)
except ValueError:
    print('Invalid quantity. Enter a whole number.')
else:
    print(f'Quantity accepted: {quantity}')
finally:
    print('Validation attempt complete.')
````

### When `pass` is appropriate
`pass` means “do nothing here.” It is occasionally useful when an invalid record should be ignored deliberately, but it should be used sparingly. In data work, log or count skipped records whenever possible so you know what was excluded.

````python
raw_values = ['10', 'unknown', '25', '', '18']
valid_values = []
skipped_values = 0

for raw_value in raw_values:
    try:
        valid_values.append(int(raw_value))
    except ValueError:
        skipped_values += 1
        pass

print('Valid values:', valid_values)
print('Skipped values:', skipped_values)
````

## Code-Along 1.2.2 — Robust numeric input validator

### Goal
Create a reusable function that accepts a text value, converts it to a number, checks an allowed range, and returns a helpful result.

### Instructor transcript
Say: *A robust program distinguishes between two questions: Can this value be converted to a number? And, if it can, is it a valid number for our business rule? A `try` block handles conversion; an `if` statement handles the rule.*

````python
def validate_score(score_text, minimum=0, maximum=100):
    """Return a validated numeric score or a clear error message."""
    try:
        score = float(score_text.strip())
    except AttributeError:
        return 'Invalid input: provide text or a value that supports strip().'
    except ValueError:
        return 'Invalid input: enter a numeric value.'
    else:
        if score < minimum or score > maximum:
            return f'Invalid score: enter a value from {minimum} through {maximum}.'
        return score

test_scores = [' 87.5 ', 'one hundred', '-2', '105', '70']

for test_score in test_scores:
    print(f'{test_score!r} -> {validate_score(test_score)!r}')
````

### Why return instead of only print?
A validation function that returns a number or message can be reused in other functions, tests, scripts, and notebooks. Printing is useful for a learner-facing interface, but returning a value makes logic easier to compose and test.

````python
def label_validated_score(score_text):
    result = validate_score(score_text)

    if isinstance(result, str):
        return result
    if result >= 70:
        return f'{result}: On track'
    return f'{result}: Needs more practice'

for value in ['84', '62', 'not available']:
    print(label_validated_score(value))
````

## Debugging checklist

When code fails, work through this sequence:

1. Read the final line of the traceback.
2. Identify the error type.
3. Find the referenced line in your code.
4. Inspect the value and its type with `print(value)` and `print(type(value))`.
5. Fix the underlying assumption rather than hiding the error.
6. Test both valid and invalid inputs.

### Common anti-patterns
- Catching every error with a bare `except:`.
- Placing too much code inside one `try` block.
- Using `pass` without tracking what was skipped.
- Treating missing dictionary keys as impossible when working with external data.
- Assuming a notebook was run top-to-bottom after restarting its kernel.
